## 3. Завантаження набору даних
У роботі використовується навчальний набір Iris із бібліотеки scikit-learn. Він містить вимірювання квіток трьох видів ірису.

```python
iris = load_iris(as_frame=True)
print(type(iris))
print(iris.keys())
```

```python
print("Назви ознак:")
for feature_name in iris.feature_names:
    print("-", feature_name)

print("\nНазви класів:")
for target_name in iris.target_names:
    print("-", target_name)
```

```python
df = iris.frame.copy()
display(df.head())
```

```python
print(df["target"].value_counts().sort_index())

target_mapping = {
    0: "setosa",
    1: "versicolor",
    2: "virginica"
}

df["species"] = df["target"].map(target_mapping)
df = df.drop(columns=["target"])

df = df.rename(
    columns={
        "sepal length (cm)": "sepal_length_cm",
        "sepal width (cm)": "sepal_width_cm",
        "petal length (cm)": "petal_length_cm",
        "petal width (cm)": "petal_width_cm"
    }
)
display(df.head())
```

## 4. Збереження даних у форматі CSV
Підготовлена таблиця зберігається на Google Drive, після чого повторно завантажується засобами бібліотеки pandas.

```python
CSV_PATH = DATA_DIR / "iris.csv"
df.to_csv(CSV_PATH, index=False)
print("Файл збережено:", CSV_PATH)

if CSV_PATH.exists():
    print("Файл успішно створено.")
    print("Розмір файла:", CSV_PATH.stat().st_size, "байт")
else:
    print("Файл не знайдено.")
```

## 6. Візуалізація даних
Гістограми використовуються для первинного дослідження розподілів числових ознак.

```python
feature_columns = [
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
]
data[feature_columns].hist(bins=15, figsize=(10, 8))
plt.suptitle("Розподіли числових ознак набору Iris")
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

### Аналіз розподілів ознак:
1. **Найменші значення:** ознака `petal_width_cm` (діапазон значень починається від 0.1 см).
2. **Найбільший діапазон:** ознака `petal_length_cm` (значення змінюються приблизно від 1.0 до 6.9 см, розмах майже 5.9 см).
3. **Форма розподілів:** ні, розподіли відрізняються. `sepal_width_cm` має розподіл, наближений до нормального (дзвоноподібний), тоді як `petal_length_cm` та `petal_width_cm` є бімодальними.
4. **Кілька груп значень:** для ознак довжини та ширини пелюстки (`petal_length_cm` і `petal_width_cm`) чітко помітні дві окремі групи значень (менші значення відповідають одному класу, більші — іншим двом).

```python
class_counts = data["species"].value_counts().sort_index()
class_counts.plot(kind="bar", figsize=(8, 5))
plt.title("Кількість об’єктів кожного класу")
plt.xlabel("Вид ірису")
plt.ylabel("Кількість об’єктів")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
for species_name, group in data.groupby("species"):
    plt.scatter(
        group["petal_length_cm"],
        group["petal_width_cm"],
        label=species_name,
        alpha=0.8
    )
plt.title("Залежність довжини та ширини пелюстки")
plt.xlabel("Довжина пелюстки, см")
plt.ylabel("Ширина пелюстки, см")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### Аналіз діаграми розсіювання:
1. **Чи утворюють окремі групи:** так, об'єкти групуються у кластери залежно від виду.
2. **Клас, що відокремлюється найкраще:** вид `setosa` лінійно відокремлюється від інших двох без жодних перекриттів (має значно менші розміри пелюсток).
3. **Класи, що частково перекриваються:** `versicolor` та `virginica` на межі дещо перекриваються за цими двома ознаками.
4. **Придатність ознак:** так, розміри пелюстки є чудовими інформативними ознаками для класифікації видів ірисів.

## 7. Постановка задачі машинного навчання
### Об'єкт дослідження
Одним об'єктом є одна квітка ірису, для якої виміряно характеристики чашолистка та пелюстки.

### Ознаки
Використовуються чотири числові ознаки:
1. довжина чашолистка (`sepal_length_cm`);
2. ширина чашолистка (`sepal_width_cm`);
3. довжина пелюстки (`petal_length_cm`);
4. ширина пелюстки (`petal_width_cm`).

### Цільова змінна
Цільовою змінною є `species` — вид квітки ірису.

### Тип задачі
Задача належить до навчання з учителем і є задачею багатокласової класифікації.

### Формулювання задачі
Необхідно за чотирма числовими характеристиками квітки автоматично визначити, до якого з трьох видів ірису вона належить: `setosa`, `versicolor` або `virginica`.

```python
X = data.drop(columns=["species"])
y = data["species"]

print("Розмір матриці ознак X:", X.shape)
print("Розмір цільової змінної y:", y.shape)

display(X.head())
display(y.head())
```

```python
expected_columns = {
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm",
    "species"
}

assert data.shape == (150, 5), f"Очікувався розмір (150, 5), отримано {data.shape}"
assert set(data.columns) == expected_columns, "Назви стовпців не відповідають очікуваним."
assert data.isna().sum().sum() == 0, "У таблиці знайдено пропущені значення."
assert data["species"].nunique() == 3, "Очікувалося три класи."
assert set(data["species"].unique()) == {
    "setosa",
    "versicolor",
    "virginica"
}, "Назви класів не відповідають очікуваним."

print("Усі основні перевірки виконано успішно.")
```
